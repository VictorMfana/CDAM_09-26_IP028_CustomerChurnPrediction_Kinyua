## 6. Check for duplicate observations
Two checks: (a) fully identical rows, (b) duplicate `Customer_ID` — a repeated ID with
different data is a different kind of problem than a true duplicate row.

```python
full_dupes = df.duplicated().sum()
print(f"Fully duplicate rows: {full_dupes}")

id_dupes = df["Customer_ID"].duplicated().sum()
print(f"Rows with a repeated Customer_ID: {id_dupes} out of {df.shape[0]}")
print(f"Unique Customer_IDs: {df['Customer_ID'].nunique()}")
```

**Finding:** 0 fully duplicate rows, but 7,101 rows share a `Customer_ID` with at least
one other row — and those rows have different Age, Gender, and account details. `Customer_ID`
was generated from a smaller pool of values and reused; it is **not** a reliable unique
customer key in this dataset. Each row is still treated as an independent customer record.
This is a data quality limitation worth stating explicitly in the report, not something we
silently "fix".

## 7. Identify missing values

```python
missing = df.isnull().sum()
```

**Finding:** `Education_Level` has 1,026 missing values (10.26%) — the only column affected.

## 8. Handle missing values
For a categorical column with ~10% missing and no obvious relationship to force a specific
value, an explicit **"Unknown"** category is safer than imputing the mode — imputing would
artificially inflate one education group and could bias the model. This also preserves the
missingness itself as potential signal.

```python
df["Education_Level"] = df["Education_Level"].fillna("Unknown")
df["Education_Level"].value_counts()
```

## 9. Check for impossible or unrealistic observations

```python
checks = {
    "Negative Age": (df["Age"] < 0).sum(),
    "Age unrealistic (<18 or >100)": ((df["Age"] < 18) | (df["Age"] > 100)).sum(),
    "Negative Account_Tenure_Months": (df["Account_Tenure_Months"] < 0).sum(),
    "Negative Average_Balance": (df["Average_Balance"] < 0).sum(),
    "Negative Monthly_Transactions": (df["Monthly_Transactions"] < 0).sum(),
    "Complaint_Resolution_Rate out of [0,1]": (
        (df["Complaint_Resolution_Rate"] < 0) | (df["Complaint_Resolution_Rate"] > 1)
    ).sum(),
    "Number_of_Products <= 0": (df["Number_of_Products"] <= 0).sum(),
    "Churn not in {0,1}": (~df["Churn"].isin([0, 1])).sum(),
}
for check, count in checks.items():
    print(f"{check}: {count}")
```

**Finding:** No impossible values found — all numeric fields fall within realistic
bounds (Age 18–70, non-negative balances, resolution rate in [0,1], Churn strictly 0/1).

# Task 2: Data Wrangling and Manipulation

## 1. Select specific variables
customer identifier plus the demographic, account, and behavioural variables are 
most relevant to a churn analysis.

```python
selected = df[[
    "Customer_ID", "Age", "Gender", "Employment_Status", "Account_Type",
    "Account_Tenure_Months", "Number_of_Products", "Average_Balance",
    "Monthly_Transactions", "Online_Banking_Usage", "Mobile_Banking_Usage",
    "Number_of_Complaints", "Complaint_Resolution_Rate", "Churn"
]].copy()
selected.head(10)
```

## 2. Filter observations based on conditions
Restricting to **established customers** (tenure over 12 months) since the new customers 
 can distort early patterns.

```python
filtered = selected[selected["Account_Tenure_Months"] > 12].copy()
print(f"Rows before filter: {selected.shape[0]}")
print(f"Rows after filter (tenure > 12 months): {filtered.shape[0]}")
```

## 3. Create at least two new variables
- `Total_Digital_Usage`: combined online + mobile banking usage for a single unit of easure
- `Balance_per_Product`: average balance spread across the number of products held which helps explain  how "deep" the relationship is per product.

```python
filtered["Total_Digital_Usage"] = filtered["Online_Banking_Usage"] + filtered["Mobile_Banking_Usage"]
filtered["Balance_per_Product"] = (filtered["Average_Balance"] / filtered["Number_of_Products"]).round(2)
filtered[["Total_Digital_Usage", "Balance_per_Product"]].describe()
```

## 4. Rename variables where appropriate
Shortening names for clarity in later tables and plots.

```python
filtered = filtered.rename(columns={
    "Account_Tenure_Months": "Tenure_Months",
    "Average_Balance": "Avg_Balance"
})
filtered.columns.tolist()
```

## 5. Sort observations
Sorting by churn status, then by balance - this will show the highest-value churned customers first.

```python
sorted_df = filtered.sort_values(by=["Churn", "Avg_Balance"], ascending=[False, False])
sorted_df.head(10)
```

## 6 & 7. Group by a categorical variable and compute summary statistics

```python
summary_table = filtered.groupby("Account_Type").agg(
    Avg_Balance=("Avg_Balance", "mean"),
    Avg_Tenure_Months=("Tenure_Months", "mean"),
    Avg_Digital_Usage=("Total_Digital_Usage", "mean"),
    Churn_Rate=("Churn", "mean"),
    Customer_Count=("Customer_ID", "count")
).round(2).sort_values("Churn_Rate", ascending=False)
summary_table
```

## 8. Clean summary table
Same table, formatted for direct use in the report (percentages and currency-style formatting).

```python
report_table = summary_table.copy()
report_table["Churn_Rate"] = (report_table["Churn_Rate"] * 100).round(1).astype(str) + "%"
report_table["Avg_Balance"] = report_table["Avg_Balance"].apply(lambda x: f"{x:,.0f}")
report_table
```

###  interpretation
- Filtering to established customers (tenure > 12 months) dropped the dataset from 10,000 to
  9,343 rows — most customers are past the one-year mark, so this filter mainly removes very
  new accounts rather than a large chunk of the base.
- **Account type shows only a small difference in churn rate** — Savings accounts churn
  slightly more (≈27%) than Fixed Deposit accounts (≈25%), but the gap is narrow. This
  suggests account type alone is a weak churn predictor on its own, and the model in Task 4
  will need to combine it with behavioural variables (complaints, digital usage, balance) to
  find stronger signal — a useful, honest finding to report rather than overselling it.
- Fixed Deposit customers carry the highest average balance and tenure, consistent with that
  product being a longer-term commitment.
- `Total_Digital_Usage` and `Balance_per_Product` are now available as engineered features for
  the EDA (Task 3) and modelling (Task 4) stages.
